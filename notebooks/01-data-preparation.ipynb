```python
import pandas as pd
```

```python
anage = pd.read_csv('anage_data.txt', sep='\t')
anage_og = anage.copy()
anage.head(10)
```

```python
anage.info()
```

```python
filtered_data = anage.dropna(subset=['Maximum longevity (yrs)', 'Adult weight (g)'])
filtered_data = filtered_data.drop(['IMR (per yr)', 'MRDT (yrs)','Source','References'], axis = 1)
filtered_data.info()
```

```python
filtered_data['Class'].value_counts()
```

```python
mammals = filtered_data[filtered_data['Class'] == 'Mammalia'].copy()
mammals.info()
```

```python
print(mammals['Data quality'].value_counts())
```

```python
mammals = mammals[~mammals['Data quality'].isin(['questionable'])]
```

```python
mammals['Order'].value_counts()
```

```python
mammals = mammals[mammals['Order'] != 'Tubulidentata'].copy()
```

```python
mammals.info()
```

```python
from sklearn.model_selection import train_test_split

train, test = train_test_split(mammals, test_size = 0.2, 
                               stratify = mammals['Order'],
                              random_state = 42)

print(train['Order'].value_counts(normalize = True).sort_index())
print(test['Order'].value_counts(normalize = True).sort_index())
```

```python
import os
os.makedirs('data/processed', exist_ok = True)

train.to_csv('data/processed/train.csv')
test.to_csv('data/processed/test.csv')
```
